```python
# What if the list is sorted? Can we search faster?
# Binary Search: O(log n)

def binary_search(arr, left, right, x):
    if right >= left:

        mid = left + (right - left) // 2

        # If the element is present at the middle
        if arr[mid] == x:
            return mid

        # If the element is smaller than the middle element,
        # search in the left subarray
        elif arr[mid] > x:
            return binary_search(arr, left, mid - 1, x)

        # Otherwise, search in the right subarray
        else:
            return binary_search(arr, mid + 1, right, x)

    # Element is not present
    return -1


arr = [10, 20, 30, 31, 40, 50]
q = 31

binary_search(arr, 0, len(arr) - 1, q)
```

```python
import math

arr = [101, 105, 110, 115, 120, 125, 130, 135, 140, 145]

def binarySearch(arr, low, high, x):

    while low <= high:
          mid = high + math.floor((low-high)/2)
          if arr[mid] == x:
              return arr[mid]
          elif arr[mid] <= x:
              return binarySearch(arr, mid+1, high, x)
          else:
              return binarySearch(arr, low, mid-1, x)
q = 110
binarySearch(arr, 0, len(arr)-1, q)
```

```python
# Find elements common in two lists:
import random
l1 = list(range(100))
random.shuffle(l1)

l2 = list(range(50))
random.shuffle(l2)

# find common elements : O(n*m)
cnt = 0
for i in l1:
    for j in l2:
        if i == j:
          print(i)
          cnt += 1
print("number of common elements:", cnt)
```

```python
# Find elements common in two lists:
l1 = list(range(100))
random.shuffle(l1)

l2 = list(range(50))
random.shuffle(l2)

# find common elemnts in lists in O(n) time and O(m) space if m<n
## add all elements in the smallest list into a hashtable/Dict: O(m) space
smalllist = {}
for ele in l2:
    smalllist[ele] = 1  # any value is OK. Key is important

# Now find common element
cnt = 0;
for i in l1:
    if smalllist.get(i) != None: # search happens in constant time.
       print(i);
       cnt += 1;
print("Number of common elements:", cnt)
```
