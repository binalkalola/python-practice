```python
def prodDigits(num):
    """
    Calculates the product of all digits of a given number.
    """
    ans = 1
    while num > 0:
          product = num % 10
          ans *= product
          num = num // 10
    return ans
num = 1234
print(prodDigits(num))
```

question 6: Write a program to print twin primes less than 1000. If two consecutive odd numbers are both prime then they are known as twin primes

Answer :using for loop range(3,1000) checking number from 3 up to
        999 number n in each turn
        is_prime1 = true check if the first number n is prime
        is_prime2 = true Check if the second number 'n + 2' is prime


```python
"""
This program identifies twin primes less than 1000.
A twin prime is a pair of primes with a difference of two.
"""
for n in range(3,1000):

    # Check if the first number 'n' is prime
    is_prime1 = True
    for i in range(2, n):
        if n % i == 0:
            is_prime1 = False
            break

    # Check if the second number 'n + 2' is prime
    is_prime2 = True
    for i in range(2, n + 2):
        if (n +2) % i == 0:
            is_prime2 =False
            break

    if is_prime1 and is_prime2:
      print(f"twin primes: ({n} , {n+2})")
```

Question 7: Write a program to implement these formulae of permutations and combinations. Number of permutations of n objects taken r at a time: p(n, r) = n! / (n-r)!. Number of combinations of n objects taken r at a time is: c(n, r) = n! / (r!*(n-r)!) = p(n,r) / r!

Answer :define factorial() function its return
        Calculates the factorial of a given number.
        
        define permutation function its return
        Calculates the number of permutations of n objects taken r at a time.

        define combination function
        Calculates the number of combinations of n objects taken r at a time.



```python
# Factorial function
def factorial(n):
  """
  Calculates the factorial of a given number.
  """
  fact = 1
  for i in range(1, n + 1):
      fact *= i
  return fact

# Permutation function
def permutation (n, r):
    """
    Calculates the number of permutations of n objects taken r at a time.
    """
    return factorial(n) // factorial(n - r)

# Combination function
def combination (n ,r):
    """
    Calculates the number of combinations of n objects taken r at a time.
    """
    return factorial(n) // (factorial(r) * factorial(n - r))

# Input
n = int(input("Enter n: "))
r = int(input("Enter r: "))

# Output
print("permutation p(n, r) =", permutation(n, r))
print("combination c(n, r) =", combination(n, r))

```

Question 8 : Write a function that converts a decimal number to binary number

Answer : define decimal_to_binary funtion its return binary
         while loop check n > 0 if you enter 0 funtion return an empty string.

         n % 2 this give the last digit of the binary(remainder)

         n // 2 floor devision its return quotient
         
         str(remainder) + binary  putting the new remainder first, you don't have to reverse the string at the end. You are building the binary number from right to left.

```python
def decimal_to_binary(n):
  """
  Converts a decimal integer to its binary representation as a string.
  """
  if n == 0:
      return "0"
  binary = ""

  while n > 0:
        remainder = n % 2
        binary = str(remainder) + binary
        n = n // 2
  return binary
n = int(input("Enter number:"))
print(decimal_to_binary(n))
```

question 9 :Write a function cubesum() that accepts an integer and returns the sum of the cubes of individual digits of that number.
Use this function to make functions PrintArmstrong() and isArmstrong() to print Armstrong numbers and to find whether is
an Armstrong number

Answer :define function cubesum() its returns the sum of
        individual digits of a number
        define funtion isArmstrong() it checks if a number is an armstrong
        return true if it is, false otherwise.
        define function printArmstrong() print all armstrong number from 0 up to given limit number

```python
def cubesum(n):
    """
    Returns the sum of the cubes of individual digits of a number.
    """
    sum = 0
    for digit in str(n):
        sum = sum + (int(digit)**3)
    return sum

def isArmstrong(n):
    """
    Checks if a number is an Armstrong number.
    Returns True if it is, False otherwise.
    """
    if cubesum(n) == n:
       return True
    else:
       return False

def printArmstong(limit):
    """
    Prints all Armstrong numbers from 0 up to the given limit.
    """
    for i in range(limit+1):
        if isArmstrong(i):
          print(i)

print("Armstrong numbers between 1 and 500:")
printArmstong(500)
```

question 10: Write a function sumPdivisors() that finds the sum of proper divisors of a number. Proper divisors of a number are those numbers by which the number is divisible, except the number itself. For example proper divisors of 36 are 1, 2, 3, 4, 6, 9, 18

Answer :define sumPdivisors() function its Calculates the sum
        of all proper divisors of a number.

        for i in range(1, num):: The computer starts a search. It picks every number starting from 1 up to num - 1

        if num % i == 0:: For every number i, the computer performs a division check.

        total += i: If the test above is successful, the computer takes that number i and adds it to our total container.

```python
def sumPdivisors(num):
    """
    Calculates the sum of all proper divisors of a number.
    """
    total = 0
    for i in range(1, num):
        if num % i == 0:
           total += i
    return total

print(sumPdivisors(36))
```

Question 11: If all digits of a number n are multiplied by each other repeating with the product, the one
digit number obtained at last is called the multiplicative digital root of n. The number of
times digits need to be multiplied to reach one digit is called the multiplicative
persistance of n.
Example: 86 -> 48 -> 32 -> 6 (MDR 6, MPersistence 3)
               341 -> 12->2           (MDR 2, MPersistence 2)
Using the function prodDigits() of previous exercise write functions MDR() and
MPersistence() that input a number and return its multiplicative digital root and
multiplicative persistence respectively

Answer : define prodDigit() function its returns the product of all
        digits of givin number n.
        difine MDR() function MDR of a number
        It repeatedly multiplies the digits of n until a single digit is obtained.
        define MPersistence() function Returns the Multiplicative Persistence of a number.
        It counts how many times digits must be multiplied to reach a single digit.

```python
def prodDigit(n):
    """
    Returns the product of all digits of a given number n.
    """
    product = 1
    while n > 0:
          digit = n % 10
          product *= digit
          n = n // 10
    return product

def MDR(n):
    """
    Returns the Multiplicative Digital Root (MDR) of a number.
    It repeatedly multiplies the digits of n until a single digit is obtained.
    """
    while n >= 10:
          n = prodDigit(n)
    return n

def MPersistence(n):
    """
    Returns the Multiplicative Persistence of a number.
    It counts how many times digits must be multiplied to reach a single digit.
    """
    count = 0
    while n >= 10:
          n = prodDigit(n)
          count += 1
    return count

n = int(input("Enter number : "))
print("Multiplicative Digit Root :" , MDR(n))
print("Multiplicative MPersistence :", MPersistence(n))
```

Question 12: A number is called perfect if the sum of proper divisors of that number is equal to the
number. For example 28 is perfect number, since 1+2+4+7+14=28. Write a program to
print all the perfect numbers in a given range.

Answer :define sumProperDivisor() function its returns the sum of
        all proper divisors of a number.

        define PrintPerfactNumber() function. Prints all perfect numbers from 1 to given limit.

```python
def sumProperDivisors(n):
    """
    Returns the sum of all proper divisors of a number n.
    """
    total = 0
    for i in range(1, n):
        if n % i == 0:
          total += i
    return total

def printPerfectNumbers(limit):
    """
    Prints all perfect numbers from 1 to given limit.
    """
    for num in range(1, limit+1):
      if sumProperDivisors(num) == num:
         print(num)
n = int(input("Enter limit: "))
print("perfact number up to", n, "are:")
printPerfectNumbers(n)

```

Question 13. Two different numbers are called amicable numbers if
             the sum of the proper divisors of each is equal to the other number. For example 220 and 284 are amicable numbers.Sum of proper divisors of 220 = 1+2+4+5+10+11+20+22+44+55+110 = 284 Sum of proper divisors of 284 = 1+2+4+71+142 = 220 Write a function to print pairs of amicable numbers in a range

Answer : define function sum_Proper_Divisor() its return the sum of
         all proper divisor of a number.
         define print_ambicale_pair()
         Prints all amicable number pairs within a given range.
         Two numbers (a, b) are amicable if:
         sum of proper divisors of a = b
         sum of proper divisors of b = a
         a ≠ b





```python
def sum_proper_divisor(n):
    """
    Calculates the sum of all proper divisors of a number.
    """
    total = 0
    for i in range(1, n):
        if n % i == 0:
           total += i
    return total
def print_amicable_pair(limit):
    """
    Prints all amicable number pairs within a given range.

    Two numbers (a, b) are amicable if:
    - sum of proper divisors of a = b
    - sum of proper divisors of b = a
    - a ≠ b
    """
    for a in range(1, limit+1):
        b = sum_proper_divisor(a)

        # avoid duplicates and self-pair
        if b > a and b <= limit:
           if sum_proper_divisor(b) == a:
              print(a,"and", b, "are amicable numbers")
print_amicable_pair(500)
```

Question 14 : Write a program which can map() to make a list   
             whose elements are cube of elements in a given list
            
Answer  :define get_cube() function its returns the cube of number
         define cube_list() function its return the cube of list
         use map() to create a list of cube of given list element.


```python
lst = [1, 2, 3, 4, 5]
def get_cube(n):
    """
    Return the cube of number
    """
    return n ** 3

def cube_list(lst):
    """
    Uses map() to create a list of cubes of given list elements.
    """
    return list(map(get_cube, lst))

result = cube_list(lst)
print(result)
```
