**Steps to follow:**

**Step 1 — Abstract base class `Employee`**
Inherit from `ABC`. Add `__init__` with `name`, `department`, and `_salary` (protected). Add `@abstractmethod calculate_salary()` that child classes must implement. Add a concrete method `print_details()` that prints one formatted line.

**Step 2 — `FullTimeEmployee(Employee)`**
`__init__` takes `name`, `department`, `monthly_salary`. Implement `calculate_salary()` — returns `monthly_salary * 12`.

**Step 3 — `PartTimeEmployee(Employee)`**

> **Rules:** Use all four pillars — ABC + `@abstractmethod`, `super()`, protected attribute `_salary`, and polymorphic `calculate_salary()` call in the loop.



Answer:

      In this experiment, an Employee Management System was created using Python OOP concepts.

      An abstract class Employee was created using ABC and @abstractmethod. It contains common details like employee name, department, and protected salary (_salary).

      Three employee classes were created by inheriting the Employee class:

      **FullTimeEmployee**:
            Calculates annual salary using monthly_salary × 12.
      **PartTimeEmployee:**
            Calculates annual salary using hourly_rate × hours_per_month × 12.
      **ContractEmployee**:
            Uses the given contract amount as yearly salary.

      Encapsulation was implemented using _salary with @property and setter to control salary values.

      Polymorphism was used by calling the same calculate_salary() method for different employee objects.

      The print_report() function displays employee details, total employees, and total payroll with proper formatting.

      Concepts Used:

        **Abstraction**: Used abstract class and abstract method.
        **Inheritance**: Child classes inherited features from the Employee class.

      Conclusion:

        The Employee Management System was successfully created using the four pillars of OOP. This experiment helped in understanding how OOP concepts make programs organized, reusable, and easy to manage.

````python
from abc import ABC, abstractmethod

class Employee(ABC):

      def __init__(self, name, department, salary):

          self.name = name
          self.department = department
          self._salary = salary


      @abstractmethod
      def calculate_salary(self):
          pass

      def print_details(self):
         #print(self.name)
         #print(self.department)
         #print(self.calculate_salary)
         #print(self.calculate_salary())
         print(f"{self.name:<12} [{self.employee_type:<9}] Salary: ₹{self.calculate_salary():<10.2f}  Dept: {self.department:<15}")

class FullTimeEmployee(Employee):
      def __init__(self, name, department, monthly_salary):
          super().__init__(name, department, monthly_salary)
          self.employee_type = "Full-Time"

      def calculate_salary(self):
          return self._salary * 12


class PartTimeEmployee(Employee):

    def __init__(self, name, department, hourly_rate, hours_per_month):

        annual_salary = hourly_rate * hours_per_month * 12

        super().__init__(name, department, annual_salary)

        self.employee_type = "Part-Time"
        self.hourly_rate = hourly_rate
        self.hours_per_month = hours_per_month

    def calculate_salary(self):
        return self._salary

class ContractEmployee(Employee):

    def __init__(self, name, department, contract_amount):

      # contract_amount = annual_salary

       super().__init__(name, department, contract_amount)

       self.employee_type = "Contract"

       #self.contract_amount = contract_amount

    def calculate_salary(self):
        return self._salary

@property
def salary(self):
    return self._salary

@salary.setter
def salary(self, value):

    if  value < 0:

        print("Salary cannot be negative!")

    else:

        self._salary = value

emp1 = FullTimeEmployee("Ravi", "Engineering", 70000.00)
emp2 = PartTimeEmployee("Priya", "Design", 150, 15)
emp3 = ContractEmployee("Arjun", "Data Science", 75000.00)
employees = [emp1, emp2, emp3]

def print_report(employees):

    print("=" * 60)
    #print("PYTHON ACADEMY — Employee Report")
    print(f"{'PYTHON ACADEMY — Employee Report':^60}")
    print("=" * 60)

    total_payroll = 0

    for emp in employees:

        emp.print_details()

        total_payroll += emp.calculate_salary()

    #print(len(employees))
    #print(total_payroll)
    print("=" * 60)
    print(f"Total employees : {len(employees)}")
    print(f"Total payroll   : ₹{total_payroll:.2f}")
    print("=" * 60)

print_report(employees)

````

````python
from abc import ABC, abstractmethod

class Vehicle(ABC):

      def __init__(self, vehicle_name, brand, daily_rent):

          self.vehicle_name = vehicle_name
          self.brand = brand
          self._daily_rent= daily_rent

      @abstractmethod
      def calculate_rent(self, days):
         pass

      def print_details(self, days):

          #print(self.vehicle_name)
          #print(self.vehicle_type)
          #print(self.brand)
          #print(self.calculate_rent(days))
          print(f"{self.vehicle_name:<10} [{self.vehicle_type:<5}] Brand: {self.brand:<9} Rent: ₹{self.calculate_rent(days):.2f}")



class Car(Vehicle):

      def __init__(self, vehicle_name, brand, daily_rent, insurance_charge):
          super().__init__(vehicle_name, brand, daily_rent)
          self.insurance_charge = insurance_charge
          self.vehicle_type = "Car"

      def calculate_rent(self, days):
          return days * self._daily_rent + self.insurance_charge

class Bike(Vehicle):

      def __init__(self, vehicle_name, brand, daily_rent, helmet_charge):
          super().__init__(vehicle_name, brand, daily_rent)
          self.helmet_charge = helmet_charge
          self.vehicle_type = "Bike"

      def calculate_rent(self, days):
          return days * self._daily_rent + self.helmet_charge

class Truck(Vehicle):

      def __init__(self, vehicle_name, brand, daily_rent, loading_charge):
          super().__init__(vehicle_name, brand, daily_rent)
          self.loading_charge = loading_charge
          self.vehicle_type = "Truck"
      def calculate_rent(self, days):
          return days * self._daily_rent + self.loading_charge

@property
def daily_rent(self):
    return self._daily_rent

@daily_rent.setter
def daily_rent(self, value):
    if value < 0:
       print("Daily rent cannot be nagative")
    else:
        self._daily_rent = value

car = Car("Swift", "Maruti", 800, 800)

bike = Bike("Activa", "Honda", 250, 150)

truck = Truck("Tata Ace", "Tata", 1200, 2000)

vehicles = [car, bike, truck]

def print_report(vehicles, days):

    print("=" * 60)
    #print("PYTHON ACADEMY — Vehicle Rental Report")
    print(f"{'PYTHON ACADEMY — Vehicle Rental Report':^60}")
    print("=" * 60)

    total_rent = 0

    for vehicle in vehicles:

        vehicle.print_details(days)

        total_rent += vehicle.calculate_rent(days)

    print("=" * 60)
    print(f"Total Vehicles : {len(vehicles)}")
    #print(len(vehicles))
    print(f"Total Rent   : ₹{total_rent:.2f}")
    #print(total_rent)
    print("=" * 60)

print_report(vehicles, 5)

````
